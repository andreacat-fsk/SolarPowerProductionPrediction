Import libraries

```python
import pandas as pd
from pathlib import Path
import os
```

```python
def import_weather(rel_path, weather):
    """
    Imports and combines hourly rainfall data from 2013 to 2024.

    Parameters:
    rel_path (str): Relative path to the folder containing yearly rainfall files.

    Returns:
    pd.DataFrame: Combined DataFrame with an added 'year' column.
    """
    years = range(2013, 2024) 
    base_path = Path(rel_path)
    all_data = []

    for year in years:
        file_path = base_path / f"{weather}_{year}.txt"
        try:
            df = pd.read_csv(file_path, header=None, 
                             sep=",", 
                             engine="python",
                             on_bad_lines="warn",
                             parse_dates=[0],
                             date_format="%Y-%m-%d %H:%M")
            df["year"] = year
            all_data.append(df)
            print("Year", year, "appended")
        except FileNotFoundError:
            print(f"Warning: File not found for year {year} at {file_path}")
        
    if not all_data:
        raise ValueError("No data files found. Check the path and file names.")

    return pd.concat(all_data, ignore_index=True)
```

## Rainfall data prep and output

We have to fix the csv files as the column number is inconsistent.

```python
def fix_rainfall_file(path, expected_cols, output_path=None):
    """
    Fixes a rainfall data file by adjusting rows to match expected number of columns.
    
    Parameters:
    - path: str, path to input file
    - expected_cols: int, expected number of comma-separated values
    - output_path: str, optional output path to save fixed file; if None, overwrites input
    """
    fixed_lines = []

    with open(path, "r", encoding="utf-8") as file:
        for line in file:
            fields = line.rstrip("\n").split(",")

            if len(fields) > expected_cols:
                fixed_line = ",".join(fields[:expected_cols])
            elif len(fields) < expected_cols:
                fixed_line = ",".join(fields + [""] * (expected_cols - len(fields)))
            else:
                fixed_line = ",".join(fields)

            fixed_lines.append(fixed_line + "\n")

    if output_path is None:
        output_path = path  

    with open(output_path, "w", encoding="utf-8") as file:
        file.writelines(fixed_lines)

    print(f"Fixed file saved to: {output_path}")

data_dir = "Data\\Weather Data\\Hourly Rainfall UK"

for year in range(2013, 2025):
    file_name = f"rainfall_{year}.txt"
    file_path = os.path.join(data_dir, file_name)

    if year in [2013, 2014]:
        expected = 15
    elif year in range(2016, 2025):
        expected = 15
    elif year == 2015:
        expected = 15  

    fix_rainfall_file(file_path, expected_cols=expected)
```
